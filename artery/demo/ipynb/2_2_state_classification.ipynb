# 2.2 State Classification

After demodulation, the integrated IQ features are classified into quantum readout states. In the software demo this is implemented with clustering and center-based classification.

```python
import numpy as np

rng = np.random.default_rng(3)
state0 = rng.normal(loc=(-1.0, 0.2), scale=0.25, size=(200, 2))
state1 = rng.normal(loc=(1.0, -0.2), scale=0.25, size=(200, 2))
center0 = state0.mean(axis=0)
center1 = state1.mean(axis=0)
print('center0:', center0)
print('center1:', center1)

def classify(point):
    return int(np.linalg.norm(point - center1) < np.linalg.norm(point - center0))

print('example classification:', classify(np.array([0.8, -0.1])))
```
