# 2.3 Feedback Prediction

ARTERY uses early measurement evidence and branch history to issue feedback before waiting for a long full readout whenever confidence is high enough.

```python
def early_decision(p_state1, threshold_hi=0.90, threshold_lo=0.10):
    if p_state1 >= threshold_hi:
        return 'branch1', True
    if p_state1 <= threshold_lo:
        return 'branch0', True
    return 'continue', False

for p in [0.04, 0.35, 0.91]:
    print(p, early_decision(p))
```

## Tutorial Interpretation

The reproduced hardware interface exposes the same key control idea:

- Accumulate trajectory evidence from the readout stream.
- Estimate confidence for state 0 or state 1.
- Exit early if the confidence threshold is reached.
- Otherwise return a final decision at the end of the allowed readout window.
