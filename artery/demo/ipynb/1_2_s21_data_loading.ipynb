# 1.2 S21 Data Loading

This notebook shows the software-side readout data format. The expected input is `s21_data.mat`, placed in the software project root when running the full demo.

```python
from pathlib import Path
import scipy.io as sio

data_path = Path('../../software/s21_data.mat')
print('Expected data path:', data_path)
print('Exists:', data_path.exists())

if data_path.exists():
    read_data = sio.loadmat(data_path)
    print(read_data.keys())
    print('data shape:', read_data['data'].shape)
else:
    print('Data file is not included in the repository. Place s21_data.mat under artery/software/.')
```
