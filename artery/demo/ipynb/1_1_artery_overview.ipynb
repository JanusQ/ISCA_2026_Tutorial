# 1.1 ARTERY Feedback Overview

This notebook introduces the reproduced ARTERY-style quantum feedback flow used in this tutorial topic.

The goal is to connect software readout analysis with an FPGA feedback interface: S21/IQ measurement data is demodulated, classified, and converted into an early feedback branch decision.

## System Flow

![Hardware feedback path](../results/artery_hardware_feedback_path.png)

The tutorial implementation is organized around this path:

1. Host prepares S21/IQ readout data.
2. FPGA receives data through UDP and buffers it in DDR.
3. ARTERY-style logic demodulates and analyzes the measurement trajectory.
4. A branch decision selects the feedback waveform.
5. The host receives metadata, latency, and selected waveform data.

## ARTERY-Inspired Blocks

The hardware-facing interface is structured around these functions:

- NCO and mixer for digital IQ demodulation.
- Trajectory analyzer for accumulating readout evidence.
- Branch history table for recent measurement context.
- Bayesian predictor for confidence-based early decision.
- Feedback selector for branch-dependent waveform output.
