# 2.1 IQ Demodulation

The first processing step converts raw time-domain I/Q samples into integrated features. This follows the readout-processing role used by ARTERY before trajectory classification and prediction.

```python
import numpy as np

def demodulate_iq(i_samples, q_samples, omega):
    t = np.arange(len(i_samples))
    cos = np.cos(omega * t)
    sin = np.sin(omega * t)
    i_acc = np.sum(i_samples * cos + q_samples * sin)
    q_acc = np.sum(q_samples * cos - i_samples * sin)
    return i_acc, q_acc

# Small synthetic example for tutorial execution without the private MAT file.
t = np.arange(256)
i = np.cos(0.08 * t) + 0.05 * np.random.default_rng(1).normal(size=t.size)
q = np.sin(0.08 * t) + 0.05 * np.random.default_rng(2).normal(size=t.size)
print('Integrated feature:', demodulate_iq(i, q, omega=0.08))
```
